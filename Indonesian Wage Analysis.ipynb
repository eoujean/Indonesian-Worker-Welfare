1. Analisis perbandingan Upah dan Faktor lain vs Pengeluaran

```python
from sklearn.impute import SimpleImputer
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns

# Menggabungkan data berdasarkan kolom 'provinsi'
df_combined = pd.merge(upah, pengeluaran, on='provinsi')
df_combined = pd.merge(df_combined, ump, on='provinsi')

# Memilih kolom yang relevan
df_combined = df_combined[['provinsi', 'upah', 'ump', 'peng']]

# Mengimputasi nilai yang hilang
imputer = SimpleImputer(strategy='mean')
df_combined[['upah', 'ump', 'peng']] = imputer.fit_transform(df_combined[['upah', 'ump', 'peng']])

# Variabel dependen
y = df_combined['peng']

# Variabel independen
X = df_combined[['upah', 'ump']]

# Menambahkan konstanta ke model (intercept)
X = sm.add_constant(X)
```

```python
# Membangun dan fitting model
model = sm.OLS(y, X).fit()
predictions = model.predict(X)

# Print out the statistics
print(model.summary())
```

```python
# Membuat scatter plot dengan garis regresi untuk Upah vs Pengeluaran
plt.figure(figsize=(10, 5))

# Scatter plot untuk Upah vs Pengeluaran
plt.subplot(1, 2, 1)
sns.scatterplot(x=df_combined['upah'], y=df_combined['peng'])
plt.plot(df_combined['upah'], model.predict(X), color='red')
plt.xlabel('Upah')
plt.ylabel('Pengeluaran')
plt.title('Upah vs Pengeluaran')

# Scatter plot untuk UMP vs Pengeluaran
plt.subplot(1, 2, 2)
sns.scatterplot(x=df_combined['ump'], y=df_combined['peng'])
plt.plot(df_combined['ump'], model.predict(X), color='red')
plt.xlabel('UMP')
plt.ylabel('Pengeluaran')
plt.title('UMP vs Pengeluaran')

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
```

```python
# Mendapatkan intercept dan koefisien dari model
intercept = model.params['const']
coef_upah = model.params['upah']
coef_ump = model.params['ump']

# Menampilkan intercept dan koefisien
print(f"Intercept: {intercept}")
print(f"Koefisien Upah: {coef_upah}")
print(f"Koefisien UMP: {coef_ump}")

# Membentuk persamaan regresi
print(f"Persamaan Regresi: pengeluaran = {intercept:.2f} + {coef_upah:.2f} * upah + {coef_ump:.2f} * UMP")
```

```python
import pmdarima as pm
from pmdarima.model_selection import train_test_split
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("upah.csv")
df
```

```python
x = df[df['provinsi'] == "JAWA TIMUR"]
y = x[['upah.2015', 'upah.2016', 'upah.2017', 'upah.2018', 'upah.2019', 'upah.2020', 'upah.2021', 'upah.2022']].values.flatten().astype(float)
```

```python

# Fungsi untuk membangun model ARIMA dan membuat prediksi
def predict_wages_arima_single(wages, years_to_predict):
    future_years = np.arange(2023, 2023 + years_to_predict)

    # Membuat dan melatih model ARIMA
    model = pm.auto_arima(wages, seasonal=False, stepwise=True, suppress_warnings=True)

    # Prediksi upah untuk tahun yang ditentukan
    future_wages = model.predict(n_periods=years_to_predict)

    # Visualisasi hasil prediksi
    years = np.arange(2015, 2023)
    plt.figure(figsize=(10, 6))
    plt.plot(years, wages, 'o', label='Data Aktual')
    plt.plot(future_years, future_wages, 'x-', label='Prediksi')
    plt.xlabel('Tahun')
    plt.ylabel('Upah')
    plt.title('Prediksi Upah untuk Jawa Timur')
    plt.legend()
    plt.grid(True)
    plt.show()

    return future_wages, future_years

# Memanggil fungsi untuk prediksi 5 tahun ke depan untuk Jawa Timur
future_wages, future_years = predict_wages_arima_single(y, 5)

# Menampilkan prediksi
print("Prediksi upah di Jawa Timur untuk 5 tahun ke depan:")
for year, wage in zip(future_years, future_wages):
    print(f"{year}: {wage}")
```

```python
# Menggabungkan hasil prediksi ke dalam DataFrame
future_years_str = [f'upah.{year}' for year in future_years]
prediksi_jawa_timur = {'provinsi': 'JAWA TIMUR'}
prediksi_jawa_timur.update({f'upah.{year}': wage for year, wage in zip(future_years, future_wages)})

# Membuat DataFrame hasil prediksi
combined_df = pd.DataFrame([prediksi_jawa_timur])

# Menampilkan DataFrame dengan prediksi
print(combined_df)
```
